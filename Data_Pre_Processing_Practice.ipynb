```python
import pandas as pd

df=pd.read_csv("students.csv")
print(df,'\n')

# for data from api etc
# df=pd.DataFrame(data)
# return first 5 rown by default or we can pass value
# print(df.head(),'\n')
# print(df.count(),'\n')

# Finding missing values

# df=df.isnull()
# print(df,'\n')
# df=df.isnull().sum()
# print(df,'\n')

# Handling missing values

# Method 1:remove rows
# This is useful when only a small number
#  of rows have missing values.
print(df.dropna(),'\n')

# Method 2:fill missing values
# df['age']=df['age'].fillna(df['age'].mean())
# df['age']=df['age'].fillna(df['age'].median())
df['age']=df['age'].fillna(df['age'].mode()[0])
print(df['age'],'\n')

# Remove duplicates

df=df.drop_duplicates()
print(df,'\n')

# Check Data types
print(df.dtypes)
```

```python
from sklearn.preprocessing import LabelEncoder

# Encoding

students = [
    {"name": "Ali", "age": 21, "gender": "Male", "marks": 85},
    {"name": "Sara", "age": 22, "gender": "Female", "marks": 92},
    {"name": "Ahmed", "age": None, "gender": "Male", "marks": 70},

]

# Label Encoding
df=pd.DataFrame(students)
print('Before Label Encoding','\n',df,'\n')

encoder=LabelEncoder()
df['gender']=encoder.fit_transform(df['gender'])
print('After Label Encoding','\n',df,'\n')


# One Hot Encoding
df=pd.DataFrame(students)
print('Before One Hot Encoding','\n',df,'\n')
df=pd.get_dummies(df,columns=['gender'],dtype=int)
print('After One Hot Encoding','\n',df,'\n')
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import MinMaxScaler

import pandas as pd
# Scaling

students = [
    {"name": "Ali", "age": 21, "gender": "Male", "marks": 85},
    {"name": "Sara", "age": 22, "gender": "Female", "marks": 92},
    {"name": "Ahmed", "age": None, "gender": "Male", "marks": 70},
]
df=pd.DataFrame(students)
print(df,'\n')

df['age']=df["age"].fillna(df['age'].mean())
print(df,'\n')

# Min Max Scaling
# Xscaled ​= ​ X − Xmin ​​/ Xmax ​− Xmin
scaler=MinMaxScaler()
df[["age", "marks"]] = scaler.fit_transform(df[["age", "marks"]])
print(df,'\n')
# Standardization
# Z = X − μ / σ
# scaler=StandardScaler()

# df[['age','marks']]=scaler.fit_transform(df[['age','marks']])
# print(df,'\n')
```
